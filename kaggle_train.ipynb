# 🚀 Alpha Algorithm: ~2B Parameter CNN-Backed GUI Vision-Action Model
### Complete Training Pipeline on Free Kaggle / Colab T4 GPU (8 GB–16 GB VRAM)

**Alpha Algorithm** is an end-to-end multimodal foundation model for operating system GUI navigation and grounding.
- **Vision Backbone:** Hierarchical ConvNeXt CNN (preserves 2D desktop geometry without token serialization artifacts)
- **Projector:** 2-layer GELU MLP with 2x2 spatial downsampling (144–256 visual tokens)
- **Language Decoder:** 4-bit NF4 Quantized Qwen2.5-1.5B with PEFT LoRA attention adapters
- **Memory Budget:** Fits comfortably in < 6 GB VRAM during training with `paged_adamw_8bit`
- **Export Footprint:** Lightweight checkpoint < 150 MB (Projector + LoRA adapters)
